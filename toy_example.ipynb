```python
import numpy as np
from DPHI import D_PHI
```

```python
# A 2D toy example demonstrating the calculation of D-PHI and CI
# This corresponds to the example shown in Figure 7
```

```python
# The same aspiration level (ai) and reservation level (ri) are used for both objectives
# That is, a = (5, 5), r = (30, 30)
ai = 5
ri = 30

# Desirability function values at the aspiration and reservation levels: c1 = D(ai), c2 = (ri)
c1 = 1
c2 = 0

# Upper limit of the desirability function values: c1 + epsilon_a = 1.2
epsilon_a = 0.2
# Lower limit of the desirability function values: c2 - epsilon_r = -0.2
epsilon_r = 0.2

# Reference point in the desirability space (same epsilon_r for all objectives)
ref_point = np.array([c2 - epsilon_r, c2 - epsilon_r])
```

```python
# A set of objective vectors (non-dominated set in the objective space)
objs = np.array([[-10,50], [0,35], [10,25], [15,20], [20,15], [25,10], [35,0],[50,-10]])
```

```python
# Initialize the indicators with the set
ind_cal = D_PHI(objs, ai, ri)
# Calculate D-PHI and CI values of the set
dphi_set, ci_set = ind_cal.get_values()

print("D-PHI value of the set: ", dphi_set)
print("CI value of the set: ", ci_set)
```

```python
# Calculate D-PHI and CI value of each objective vector in the set
D = ind_cal.df()  # Desirability function value of objs

# The difference between desirability function values and the lower limit (-epsilon_r)
diff1 = D[:, 0]  - ref_point[0]
diff2 = D[:, 1] - ref_point[1]
# Compute D-PHI values of each objective vector, (D1 - epsilon_r) * (D2 - epsilon_r)
dphi_each = diff1 * diff2

# Compute ASF values of each objective vector
asf_each = ind_cal.ASF_value()
index_asf = np.argmin(asf_each)

print("D-PHI value of each point: ", dphi_each)
print("Index of the point determining the CI value: ", index_asf + 1)
```
